```python
import json
from pathlib import Path

proxy_path = Path("/content/sec_10k_table_proxies.jsonl")

proxies = [
    json.loads(line)
    for line in proxy_path.read_text(encoding="utf-8").splitlines()
    if line.strip()
]

print(f"Proxies: {len(proxies)}")

for proxy in proxies:
    print("\n" + "=" * 80)
    print(proxy["proxy_text"])
```
