```python
import json
from pathlib import Path

chunk_path = Path("c:/Users/nikola.bakic/OneDrive - Sixsentix AG/Documents/Repositories/Banking-Technology-and-Operational-Risk-Intelligence-Assistant/data/processed/chunks/sec_10k_chunks.jsonl")

chunks = [
    json.loads(line)
    for line in chunk_path.read_text(encoding="utf-8").splitlines()
    if line.strip()
]

standalone_numbers = [
    chunk
    for chunk in chunks
    if (
        chunk["element_type"] == "text"
        and chunk["text"].strip().isdigit()
        and len(chunk["text"].strip()) <= 3
    )
]

chunks_with_pages = [
    chunk
    for chunk in chunks
    if chunk["page_start"] is not None
]

cross_page_chunks = [
    chunk
    for chunk in chunks_with_pages
    if chunk["page_start"] != chunk["page_end"]
]

print(f"Ukupno chunkova: {len(chunks)}")
print(f"Chunkovi sa stranicom: {len(chunks_with_pages)}")
print(f"Chunkovi preko više stranica: {len(cross_page_chunks)}")
print(f"Standalone brojevi u sadržaju: {len(standalone_numbers)}")

for chunk in cross_page_chunks[:5]:
    print(
        chunk["ticker"],
        f"{chunk['page_start']}-{chunk['page_end']}",
        repr(chunk["text"][:150]),
    )
```
